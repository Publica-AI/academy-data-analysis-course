## 1. Recap: variables and data types

A Lagos retail chain scenario runs through this whole notebook. Start by storing a few
facts about one branch as variables.

````python
branch_name  = "Lagos"
total_sales  = 245000
average_sale = 15312.50
target_met   = True

print(branch_name, total_sales, average_sale, target_met)
````

- `str` — text labels, e.g. `"Lagos"`
- `int` / `float` — numbers you calculate with
- `bool` — a `True` or `False` check

**Excel parallel:** a Python variable is a named cell. `total_sales = 245000` stores a value
the way typing `245000` into a cell does, except the name can be reused anywhere in your code.

**Why an analyst uses it:** store a calculated result once, then reuse it in every later step
without retyping or recalculating it.

## 2. Lists and dictionaries in data-analysis contexts

A **list** holds a collection of values, e.g. one branch's sales across four days.

````python
daily_sales = [15000, 9500, 21000, 18000]
daily_sales
````

A **dictionary** holds one structured record with named fields, the way one spreadsheet row
has named columns.

````python
transaction = {
    "branch": "Lagos",
    "amount": "15,000",
    "date": "2026-03-01"
}

print(transaction["branch"])
````

**Why an analyst uses it:** a list of dictionaries is exactly what a small dataset looks like
before you load it into pandas next session.

## 3. Conditions and loops

Apply a business rule to every value in a list: flag each day's sales as above or below
a target.

````python
target = 12000
for sale in daily_sales:
    if sale >= target:
        print(sale, "Above target")
    else:
        print(sale, "Below target")
````

**Excel parallel:** `if` is `IF()`. Looping is filling a formula down every row, except
Python applies it in one instruction, not one drag.

**Why an analyst uses it:** flag every record against a target, threshold or category
across a whole dataset, not just the row in front of you.

## 4. Importing libraries

Analysts rarely calculate statistics by hand. Import a trusted library instead.

````python
import statistics

statistics.mean(daily_sales)
````

**Coming next:** you will `import pandas`, the library built specifically for tables of data.

## 5. Calling functions

Package the calculation above so it can be reused on any list, for any branch.

````python
def average_sale(sales_list):
    return statistics.mean(sales_list)

average_sale(daily_sales)
````

**Excel parallel:** the same idea as naming a formula once, then referencing that name
everywhere instead of rebuilding it in every cell.

This clean, well-behaved function is about to meet real, messy data. That is next.

## 6. Reading Python error messages

| Error | What it usually means |
|---|---|
| `NameError` | A variable or function name was never defined. Check the spelling. |
| `TypeError` | Two incompatible types were mixed, e.g. text and a number. |
| `ValueError` | The right type, but a value Python cannot use, e.g. text with a currency symbol passed to `float()`. |
| `KeyError` | The dictionary does not contain the key you asked for. |
| `IndexError` | You asked for a list position that does not exist. |

**Key habit:** read the last line of the traceback first, it names the problem. Then look
upward to find the line that caused it. An error is Python telling you exactly what went
wrong, not a verdict on you.

## 7. Debug it: before and after

### Before — the broken cell

This function works on a plain number string, but real branch data comes with a currency
symbol and a comma. Run the cell below and read the error it produces. **Do not panic — this
is expected.**

````python
# BROKEN — run this and read the error
def clean_amount(value):
    return float(value)

clean_amount("₦15,000")
````

You should see something like:

```text
ValueError: could not convert string to float: '₦15,000'
```

- **Means:** `float()` needs digits only. The ₦ symbol and comma are not valid parts of a number.
- **Locate it:** the input still carries currency formatting from the source file.

### After — the fix

Strip the currency symbol and the comma before converting to a number.

````python
def clean_amount(value):
    value = value.replace("₦", "")
    value = value.replace(",", "")
    return float(value)

clean_amount("₦15,000")
````

**Verify:** ₦15,000 in the source data now matches `15000.0` in the notebook, the amount
is unchanged, only its format has.

> Do not panic at an error. Read it, locate the line, fix one thing, rerun, and verify
> against a known value.
