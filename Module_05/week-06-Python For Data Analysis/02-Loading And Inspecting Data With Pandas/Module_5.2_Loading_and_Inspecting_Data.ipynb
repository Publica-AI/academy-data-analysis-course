## 1. What is pandas?

pandas is Python's library for working with tables of data: rows and columns, just
like a spreadsheet. It turns the lists of dictionaries you built in Module 5.1 into
one structured table you can sort, filter, summarise and clean in a few lines.

> "A pandas DataFrame is like an Excel worksheet that Python can manipulate
> programmatically."

## 3. Loading a CSV

`pd.read_csv(...)` opens the named file, reads every row and column into memory,
and stores the whole table in the variable you assign it to.

```python
df = pd.read_csv("Greenland_branch_sales.csv")
df
```

**Excel parallel:** this is the Python equivalent of double-clicking a CSV to open
it in Excel, except the result is a variable you can compute on, not a window you
look at.

## 4. Understanding the DataFrame

- **Columns** — the named fields across the top (`Branch`, `Product`, `Sales`, ...).
- **Index** — the row labels down the left (`0, 1, 2, ...` by default).
- **Row** — one full record, e.g. one transaction.
- **Value** — one cell, e.g. the Sales figure for one transaction.

Look at the table `df` printed above: rows and an index down the left, named
columns across the top, one value in every cell, exactly like a worksheet.

## 6. Data types

```python
df.dtypes
```

- `int64` — whole numbers, e.g. `TransactionID`, `Sales`.
- `float64` — numbers with decimals, or a numeric column that contains missing
  values, e.g. `Price`.
- `object` — text, e.g. `Branch`, `Product`, `Date`.
- `bool` — a `True`/`False` check, as in Module 5.1's `target_met`.

**Why it matters:** pandas cannot average a column it thinks is text, and a column
with even one missing value often becomes `float64` instead of `int64`. Both are
diagnostic clues, not accidents.

## 7. Missing values

```python
df.isna().sum()
```

- `isna()` marks every missing cell `True`, everything else `False`.
- `sum()` adds the `True`s down each column, one number per column.

**Why analysts check this first:** a missing price left unnoticed can silently
break a calculation later. This is exactly the check that would have caught
Module 5.1's currency-formatting error before it became one. Run it before any
analysis, not after something looks wrong.

## 8. Selecting columns

One column name in single brackets returns a **Series**.

```python
df["Sales"]
```

A list of column names in double brackets returns a **DataFrame**.

```python
df[["Sales", "Profit"]]
```

**The difference:** single brackets with one name return a Series, a single
labelled column, like clicking one column header in Excel. Double brackets with a
list of names return a DataFrame, a smaller table, like ctrl-selecting several
columns and copying them to a new sheet.

## 9. Filtering rows

```python
df[df["Sales"] > 100000]
```

- `df["Sales"] > 100000` checks every row and produces `True` or `False`.
- Wrapping it in `df[...]` keeps only the rows marked `True`.

**Excel parallel:** the same result as clicking Excel's filter arrow and keeping
only Sales greater than 100,000, except this filter is one line you can rerun on
next month's file instantly.

Combine conditions with `&` (and) or `|` (or), and wrap each condition in its own
parentheses.

## 10. Asking analytical questions

Every one of these is a business question first, and a line of pandas second.

**Which customers have sales above ₦150,000?**

```python
df[df["Sales"] > 150000]["Customer"]
```

**Which transactions occurred in the North West region?**

```python
df[df["Region"] == "North West"]
```

**Which products have missing prices?**

```python
df[df["Price"].isna()]["Product"]
```

## 11. Excel → pandas translation

| Excel | pandas |
|---|---|
| Open CSV | `pd.read_csv()` |
| Filter | Boolean filtering, e.g. `df[df["Sales"] > 100000]` |
| View first rows | `.head()` |
| Count rows/columns | `.shape` |
| Inspect columns | `.info()` |
| Check missing values | `.isna()` |

Nothing here is a new idea, only a new way of writing something you already know
how to do in Excel.
