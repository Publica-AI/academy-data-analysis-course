## Topic 2 — Indentation
Python uses spaces (indentation) to show which lines belong together, instead of `{ }` like some other languages.

Rule: lines that are indented **the same amount** and sit right under an `if` belong to that `if`.


```python
score = 78

if score >= 50:
    print("Result: PASS")     # indented -> belongs to the if
    print("Well done.")       # indented -> also belongs to the if

print("Check complete.")      # not indented -> runs no matter what
```

## Operators — Arithmetic
These do the maths. The two easy to forget:
- `/` always gives a decimal answer, even if it divides evenly.
- `//` gives the whole number part only (it "floors" the result).
- `%` gives the remainder.


```python
print(17 / 5)    # true division -> 3.4
print(17 // 5)   # floor division -> 3
print(17 % 5)    # remainder -> 2
print(2 ** 3)    # power -> 8
```

---
## Operators — Comparison
These ask a yes/no question and give back `True` or `False`.


```python
age = 30
print(age > 18)
print(age == 30)
print(age != 30)
```

---
## Operators — Logical (`and`, `or`, `not`)
Use these to combine more than one yes/no question. Write the **full comparison** on each side.


```python
age = 30

if age > 6 and age < 60:
    print("Eligible")
```

---
## Control Flow — if / elif / else
- `if` runs a block only when its condition is `True`.
- `elif` ("else if") checks another condition if the first one was `False`.
- `else` catches everything else.


```python
score = 40

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
else:
    grade = "F"

print("Grade:", grade)
```

## Loops — `while`
Use `while` when you don't know in advance how many times you'll repeat. You must update the
variable inside the loop yourself, or it will run forever.


```python
count = 1
while count <= 5:
    print(count)
    count += 1   # without this line the loop never stops
```

---
## Loops — `range()`
`range(start, stop)` counts from `start` up to (but **not including**) `stop`.


```python
for i in range(1, 6):
    print(i)   # 1, 2, 3, 4, 5 — 6 is never printed
```

---
## Loops — `for` over a list
Use `for` when you're going through a known collection of items.


```python
items = ["apple", "banana", "cherry"]

for item in items:
    print(item)
```

---
## Loops — `enumerate()`
`enumerate()` gives you both the position (starting at 0) and the item, at the same time.


```python
items = ["apple", "banana", "cherry"]

for position, item in enumerate(items):
    print(position, item)
```

## Strings — Indexing
`s[0]` is the first character. `s[-1]` is the last character.


```python
word = "Python"
print(word[0])
print(word[-1])
```

---
## Strings — Slicing
`s[start:stop]` gives you a piece of the string, from `start` up to (not including) `stop`.


```python
word = "Python"
print(word[0:3])   # first 3 characters
print(word[::-1])  # the whole string, reversed
```

---
## Strings — Immutability
Strings can't be changed in place. A method like `.strip()` gives back a **new** string —
it doesn't change the original. You have to save the result yourself.


```python
text = "  Data Science  "

text.strip()                  # the result is thrown away here
print("Unchanged:", repr(text))

text = text.strip()           # this time we save it back into text
print("Changed  :", repr(text))
```

---
## Strings — Useful Methods


```python
text = "Data Science"
print(text.upper())
print(text.lower())
print(text.replace("Data", "Computer"))
```

## Strings — Splitting and Joining


```python
sentence = "Python is readable"
words = sentence.split()      # turns the sentence into a list of words
print(words)

joined = "-".join(words)      # turns the list back into one string, joined by "-"
print(joined)
```

## Topic 8 — Lists: Creating and Accessing
A list is an ordered, changeable collection. Positions start at `0`.


```python
scores = [82, 91, 74, 88]

print("Whole list:", scores)
print("First item:", scores[0])
print("Last item :", scores[-1])
print("A slice   :", scores[1:3])
```

---
## Lists — Adding Items


```python
scores = [82, 91, 74, 88]

scores.append(95)        # adds to the end
scores.insert(0, 60)     # adds at a chosen position
print(scores)
```

---
## Lists — Removing Items
This is where most beginners slip up:
- `.remove(value)` removes by **value**.
- `.pop(index)` removes by **position**, and also gives you back what it removed.


```python
scores = [60, 82, 91, 74, 88, 95]

scores.remove(74)          # removes the value 74
print("After remove:", scores)

dropped = scores.pop(0)    # removes whatever is at position 0
print("After pop   :", scores)
print("What was dropped:", dropped)
```

---
## Lists — Sorting and Simple Maths


```python
scores = [88, 60, 82, 91, 95]

scores.sort()
print("Sorted:", scores)

print("Count  :", len(scores))
print("Sum    :", sum(scores))
print("Max    :", max(scores))
print("Average:", sum(scores) / len(scores))
```

## Tuples
A tuple looks like a list but **cannot be changed** after it's created. Use it for values that
should stay fixed, like a coordinate or a fixed-size record.


```python
record = ("Ada", 22, "Data Science")
name, age, course = record   # this is called "unpacking"

print(name)
print(age)
print(course)
```

A one-item tuple needs a comma, or Python won't treat it as a tuple.


```python
single = ("only",)
not_a_tuple = ("only")

print(type(single))
print(type(not_a_tuple))
```

## Topic 9 — Sets
A set holds only **unique** values, in no particular order. Great for removing duplicates.


```python
raw = [3, 1, 2, 3, 1, 4]
print("List:", raw)
print("Set :", set(raw))   # duplicates are gone
```

Sets also support union, intersection, and difference.


```python
python_devs = {"Ada", "Kofi", "Mary", "Yaw"}
java_devs = {"Kofi", "Mary", "Grace"}

print("In either group   :", python_devs | java_devs)
print("In both groups    :", python_devs & java_devs)
print("Python only, not Java:", python_devs - java_devs)
```

## Dictionaries — Creating and Reading
A dictionary stores labelled values, accessed by key instead of by position. Think of it as
one record — like one row of a spreadsheet.


```python
student = {"name": "Ada", "age": 22, "course": "Data Science"}

print(student["name"])
print(student.get("email", "not provided"))   # .get() avoids an error if the key is missing
```

---
## Dictionaries — Adding, Updating, Removing


```python
student = {"name": "Ada", "age": 22, "course": "Data Science"}

student["email"] = "ada@example.com"   # add a new key
student["age"] = 23                    # update an existing key
del student["course"]                  # remove a key

print(student)
```

---
## Dictionaries — Looping Through
Looping directly over a dictionary gives you the **keys** only. Use `.items()` to get
both the key and the value together.


```python
student = {"name": "Ada", "age": 23, "email": "ada@example.com"}

for key, value in student.items():
    print(key, ":", value)
```

## Topic 10 — Functions: Defining and Calling
`def` **defines** a function. Using its name with `()` **calls** it. Without the `()`
you're just pointing at the function itself, not running it.


```python
def greet(name):
    print("Hello,", name)

greet("Ada")
```

---
## Functions — `return` vs `print`
`print()` only displays a value on the screen. `return` hands a value back so you can use it later.
A function with no `return` gives back `None`.


```python
def shows(n):
    print(n * 2)      # just displays it

def gives(n):
    return n * 2       # hands it back

a = shows(5)
b = gives(5)

print("shows() gave back:", a)   # None — it never used return
print("gives() gave back:", b)   # 10
```

---
## Functions — Default Arguments
A default argument is used automatically if you don't provide your own.


```python
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Ada"))                  # uses the default greeting
print(greet("Kofi", "Good morning")) # overrides it
```

---
## Functions — Scope
A variable created **inside** a function stays inside that function. It doesn't affect
a variable with the same name outside.


```python
counter = 10   # this is the "global" counter, outside any function

def demo():
    counter = 99   # this is a brand NEW, separate variable, local to demo()
    print("Inside the function:", counter)

demo()
print("Outside the function:", counter)   # still 10
```
