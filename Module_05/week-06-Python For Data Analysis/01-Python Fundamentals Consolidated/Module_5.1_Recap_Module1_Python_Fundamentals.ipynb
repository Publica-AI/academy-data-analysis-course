## Topic 3 — Variables & Data Types

**What you covered:** variables as named containers; `snake_case` naming; `int`, `float`, `str`, `bool`;
`type()`; casting with `int()`, `float()`, `str()`.

- `=` **assigns** a value · `==` **compares** two values.

## Topic 4 — Operators & Expressions

**What you covered:** arithmetic `+ - * / // % **`; assignment `+= -= *= /=`;
comparison `== != < > <= >=`; logical `and or not`; operator precedence.

- `/` **always returns a float** — `10 / 2` is `5.0`, not `5`.
- `//` floors the result; `%` gives the remainder (the even/odd test).
- **Precedence:** `()` → `**` → `* / // %` → `+ -` → comparisons → `not` → `and` → `or`

```python
a, b = 17, 5

print("a + b  =", a + b)
print("a - b  =", a - b)
print("a * b  =", a * b)
print("a / b  =", a / b,  "  <- always a float")
print("a // b =", a // b, "  <- floor division")
print("a % b  =", a % b,  "  <- remainder")
print("a ** 2 =", a ** 2)

# % is how you test for even/odd
for n in (24, 37):
    print(n, "is even" if n % 2 == 0 else "is odd")
```

```python
# Logical operators — and / or / not
age     = 22
has_id  = True
banned  = False

can_enter = age >= 18 and has_id and not banned
print("Can enter:", can_enter)

# A very common trap:
x = 3
print("Wrong :", x == 5 or 6)        # always True — 6 on its own is truthy
print("Right :", x == 5 or x == 6)   # a full comparison on each side
```

## Topic 5 — Control Flow: Conditional Statements

**What you covered:** `if`, `if/else`, `if/elif/else`, truthy and falsy values.

- The **colon** and the **4-space indent** are both required.
- With `elif`, the **first matching condition wins** — the rest are skipped.
- **Falsy:** `0`, `""`, `[]`, `{}`, `None`. Everything else is truthy.

```python
# if / elif / else — first match wins
def classify(score):
    if score >= 90:
        return "A — Distinction"
    elif score >= 80:
        return "B — Very good"
    elif score >= 50:
        return "C — Pass"
    else:
        return "F — Fail"

for s in (95, 83, 61, 34):
    print(s, "->", classify(s))
```

```python
# Truthy / falsy
for value in (0, 1, "", "text", [], [1, 2], None, {}):
    print(f"{repr(value):<8} -> {bool(value)}")

# This is why the idiomatic empty-check reads so cleanly:
basket = []
if basket:
    print("Basket has items.")
else:
    print("Basket is empty.")        # <- this runs
```

## Topic 6 — Control Flow: Loops

**What you covered:** `while`, `for`, `range()`, `break`, `continue`, `enumerate()`.

- `while` — use when you **do not know** how many repetitions you need. **Always update the loop variable**
  or you create an infinite loop.
- `for` — use when iterating a known sequence.
- `range(stop)` · `range(start, stop)` · `range(start, stop, step)` — **`stop` is excluded**.
  `range(10)` gives `0–9`.

```python
# for + range — note the exclusive stop
print("range(5)      ->", list(range(5)))
print("range(1, 6)   ->", list(range(1, 6)))
print("range(0, 21, 5) ->", list(range(0, 21, 5)))

# The equivalent while loop — the counter MUST be updated
count = 1
while count <= 5:
    print("count =", count)
    count += 1          # forget this line and the cell never finishes
```

```python
# break, continue and enumerate
subjects = ["Maths", "Biology", "Python", "History"]

for index, subject in enumerate(subjects, start=1):
    print(index, subject)

print()
for n in range(1, 21):
    if n % 3 != 0:
        continue        # skip anything not divisible by 3
    if n > 12:
        break           # stop entirely once past 12
    print("Divisible by 3:", n)
```

## Topic 7 — Strings

**What you covered:** indexing, slicing, immutability, string methods, f-strings.

- `s[0]` is the first character, `s[-1]` the last. Slicing is `s[start:stop:step]`; `s[::-1]` reverses.
- **Strings are immutable** — a method never changes the original, so you must
  **reassign the result**: `text = text.strip()`.

```python
language = "Python"

print("First      :", language[0])
print("Last       :", language[-1])
print("Slice [1:4]:", language[1:4])     # yth — stop excluded
print("Reversed   :", language[::-1])

email = "ada.lovelace@example.com"
at = email.find("@")
print("Username   :", email[:at])
print("Domain     :", email[at + 1:])
```

```python
# Immutability — the single most common string mistake
text = "  Data Science  "

text.strip()                     # result thrown away!
print("Unchanged:", repr(text))

text = text.strip()              # captured — this is the fix
print("Changed  :", repr(text))

# Methods and f-strings
print(text.upper(), "|", text.lower(), "|", text.replace("Data", "Computer"))

words = "Python is readable".split()
print(words, "->", "-".join(words))

name, score = "Ada", 87.5678
print(f"{name} scored {score:.2f}%")     # f-string with 2-decimal formatting
```

## Topic 8 — Lists & Tuples

**What you covered:** creating and accessing lists, modifying them, iterating, tuples and immutability.

- **List** `[a, b, c]` — mutable, ordered, zero-indexed.
- `.pop(i)` takes an **INDEX** · `.remove(v)` takes a **VALUE**. Mixing these up is the classic list bug.
- **Tuple** `(a, b, c)` — immutable. A one-item tuple needs the comma: `(x,)`.

```python
scores = [82, 91, 74, 88]

print("Original :", scores)
print("First    :", scores[0], "| Last:", scores[-1])
print("Slice    :", scores[1:3])

scores.append(95)          # add to the end
scores.insert(0, 60)       # add at a position
print("After adds:", scores)

scores.remove(74)          # remove BY VALUE
dropped = scores.pop(0)    # remove BY INDEX, and return it
print("After removes:", scores, "| dropped:", dropped)

scores.sort()
print("Sorted   :", scores)
print(f"count={len(scores)} sum={sum(scores)} max={max(scores)} avg={sum(scores)/len(scores):.2f}")
```

```python
# Tuples — immutable, and unpacking
record = ("Ada", 22, "Data Science")
name, age, course = record          # unpacking
print(name, age, course)

single = ("only",)                  # the comma is what makes it a tuple
print(type(single).__name__, "vs", type(("only")).__name__)

try:
    record[0] = "Grace"             # not allowed
except TypeError as e:
    print("TypeError:", e)
```

## Topic 9 — Sets & Dictionaries

**What you covered:** sets and uniqueness, set operations, dictionaries, modifying and iterating them.

- **Set** `{a, b, c}` — unique and unordered. `set()` creates an empty set; `{}` creates a **dictionary**.
- **Dictionary** `{"key": value}` — labelled records accessed by key. This is the closest structure yet
  to a **row of a spreadsheet**, which matters for what comes next.
- Careful: `for x in my_dict` iterates the **KEYS**, not the values.

```python
# Sets remove duplicates instantly
raw = [3, 1, 2, 3, 1, 4]
print("List:", raw, "-> Set:", set(raw))

python_devs = {"Ada", "Kofi", "Mary", "Yaw"}
java_devs   = {"Kofi", "Mary", "Grace"}

print("Union       :", python_devs | java_devs)
print("Intersection:", python_devs & java_devs)
print("Difference  :", python_devs - java_devs)
print("Symmetric   :", python_devs ^ java_devs)
```

```python
# A dictionary is one labelled record
student = {"name": "Ada", "age": 22, "course": "Data Science"}

print("By key   :", student["name"])
print("Safe get :", student.get("email", "not provided"))   # no KeyError

student["email"] = "ada@example.com"   # add
student["age"] = 23                    # update
del student["course"]                  # remove

print("\nKeys  :", list(student.keys()))
print("Values:", list(student.values()))

print("\nIterating with .items():")
for key, value in student.items():
    print(f"  {key:<6} : {value}")
```

## Topic 10 — Functions

**What you covered:** defining and calling, parameters vs arguments, return values,
default arguments, and variable scope.

- `def name():` **defines**; `name()` **calls**. Without the `()` you only print the function object.
- **Parameters** appear in the definition; **arguments** are the real values you pass in.
- `return` hands a value **back**; `print` only displays it. A function with no `return` gives back `None`.
- Default arguments must come **after** all non-default ones.

```python
def average(numbers):
    """Return the mean of a list of numbers."""
    return sum(numbers) / len(numbers)

result = average([82, 91, 74])
print("Returned value:", round(result, 2))

# return vs print — the distinction that trips people up
def shows(n):
    print(n * 2)        # displays, gives back None

def gives(n):
    return n * 2        # hands the value back

a = shows(5)
b = gives(5)
print("shows() returned:", a)      # None
print("gives() returned:", b)      # 10

print("\nWithout parentheses:", gives)   # the function object, not a result
```

```python
# Default arguments and scope
def greet(name, greeting="Hello"):
    return f"{greeting}, {name}!"

print(greet("Ada"))                  # uses the default
print(greet("Kofi", "Good morning")) # overrides it

# Scope — a local name does not leak out
counter = 10                 # global

def demo():
    counter = 99             # a NEW local variable
    print("Inside :", counter)

demo()
print("Outside:", counter)   # still 10
```

## The mistakes that cost the most marks

Run this cell — each block shows the error, then the fix.

```python
# 1. input() treated as a number
raw = "25"
print("Wrong:", raw * 2)          # '2525' — string repetition
print("Right:", int(raw) * 2)     # 50

# 2. range() off-by-one
print("range(1, 10) last value:", list(range(1, 10))[-1])   # 9, not 10

# 3. .pop() vs .remove()
items = ["a", "b", "c"]
items.pop(1)          # by INDEX
print("After pop(1)   :", items)
items.remove("c")     # by VALUE
print("After remove(c):", items)

# 4. String method result not captured
s = "  spaced  "
s.strip()
print("Not captured:", repr(s))
s = s.strip()
print("Captured    :", repr(s))

# 5. Function called without parentheses
def total(): return 42
print("No parens:", type(total).__name__, "| With parens:", total())
```

## Where this is going

Everything above still applies once we move to data analysis — pandas simply does it at scale.

| What you know | What it becomes |
|---------------|-----------------|
| `list` | a column of values — a pandas **Series** |
| `dictionary` | one record — a single **row** |
| list of dictionaries | a whole table — a **DataFrame** |
| `for` loop + `if` | **filtering** rows by a condition |
| function | a reusable **transformation** |
| f-string | formatted **reporting** of results |

```python
# A table, built only from what you already know
students = [
    {"name": "Ada",   "score": 91},
    {"name": "Kofi",  "score": 74},
    {"name": "Mary",  "score": 88},
    {"name": "Yaw",   "score": 62},
]

# Filter + report — a for loop and an if, exactly as in Topics 6 and 5
passing = [s for s in students if s["score"] >= 75]

print("Passing students:")
for s in passing:
    print(f"  {s['name']:<6} {s['score']}")

average_score = sum(s["score"] for s in students) / len(students)
print(f"\nClass average: {average_score:.1f}")

# In Topic 5.2 this same structure becomes one line:
#     df = pd.DataFrame(students)
#     df[df["score"] >= 75]
```
