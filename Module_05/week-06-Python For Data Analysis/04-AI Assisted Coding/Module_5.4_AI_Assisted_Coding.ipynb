## Part 1 — What AI can and cannot do

**AI can:** generate code, explain code, suggest alternatives, debug errors,
help interpret syntax.

**AI cannot replace:** analytical judgment, data validation, output
verification, understanding.

This module isn't about avoiding AI. It's about staying the one who's
responsible for the answer.

## Part 2 — The AI-assisted coding workflow

**Define problem → Write prompt → Generate code → Read code → Run code →
Check output → Verify result → Explain code → Submit**

Copying code you can't explain is not submission-ready, no matter how correct
the output looks.

## Part 3 — Weak vs strong prompts

**Weak:** "Write pandas code." No dataset, no columns, no goal. AI has to
guess everything.

**Strong:**

> I have a pandas DataFrame `df` from branch_sales.csv with columns: Branch,
> Region, Sales, Cost, Profit, Price. Write code that filters transactions
> where Sales is above 100000, groups the result by Region, and shows total
> Sales and total Profit per region. Use only boolean filtering, groupby and
> sum. I expect a small table with 3 rows, one per region.

A strong prompt gives: dataset structure, column names, analytical objective,
expected output, and constraints.

## Part 4 — Prompting AI for pandas

**User prompt:** "My DataFrame df has columns Branch, Region, Sales, Profit.
Write pandas code to filter transactions with Sales above 100000, then group
by Region and sum Sales and Profit."

**AI-generated code:**

```python
high_value = df[df["Sales"] > 100000]
result = high_value.groupby("Region")[["Sales","Profit"]].sum()
result
```

**Human review:** Sales, Profit and Region are real columns in `df.columns`.
`Sales > 100000` matches "high value" as asked. Grouping by Region matches
the question, not Branch. The code above is the final, verified version.

## Part 5 — Ask AI to explain code

**Prompt:** "Explain this pandas code line by line as if I am a beginner data
analyst:" (the two lines above)

**AI's explanation:**

| Code | Explanation |

| `high_value = df[df["Sales"] > 100000]` | Keeps only the rows where Sales is more than 100,000. |
| `.groupby("Region")[["Sales","Profit"]].sum()` | Groups those rows by Region, adds up Sales and Profit within each group. |
| `result` | Displays the final summary table, one row per region. |

## Part 6 — AI-assisted debugging

**The broken cell.** Run it and read the real error.

```python
# BROKEN — run this and read the error
df.groupby("region")["Sales"].sum()
```

You should see `KeyError: 'region'`.

**Prompt sent to AI:** "I got this error: KeyError: 'region', from running
`df.groupby("region")["Sales"].sum()` on a DataFrame whose columns are
Branch, Region, Sales, Cost, Profit, Price. What's wrong?"

**AI's correction:** pandas column names are case-sensitive. The DataFrame
has a column called "Region" (capital R), not "region".

**Human verification, final working code:**

```python
df.groupby("Region")["Sales"].sum()
```

Send the full error message and the exact code, not "it doesn't work".
Vague reports get vague answers.

## Part 8 — AI hallucinations and incorrect code

AI can produce incorrect syntax, non-existent functions, wrong column names,
incorrect assumptions, or logically incorrect analysis, all delivered in the
same confident tone as correct code.

**A function that doesn't exist.** Run it and read the error.

```python
# BROKEN — an old pandas method AI sometimes still suggests
df.sort("Sales", ascending=False)
```

You should see `AttributeError: 'DataFrame' object has no attribute 'sort'`.
AI recalled an old pandas method removed years ago. The fix is
`sort_values()`, taught in Module 5.3: `df.sort_values("Sales",
ascending=False)`.

**A column that doesn't exist.** Run it and read the error.

```python
# BROKEN — AI guessed a plausible-sounding column name
df["Amount"].sum()
```

You should see `KeyError: 'Amount'`. AI guessed a common column name from
other datasets it has seen. This dataset calls it `Sales`.

Both mistakes are caught the same way: run the code, read the message, and
check `df.columns` before trusting any column name AI suggests.

```python
df.columns
```

**A logical error that runs fine, the most dangerous kind:** no error, no traceback, but a silently wrong answer.

```python
# PLAUSIBLE BUT WRONG — run this, it will not raise an error
df["Margin_%"] = df["Profit"] / df["Sales"] * 100
avg_margin = df["Margin_%"].sum()
print(avg_margin)
```

An average per-transaction margin of 300% is impossible. `sum()` added every row's margin instead of averaging them. Only checking whether the output makes sense catches this, the code itself never complains.

```python
# fixed
avg_margin_fixed = df["Margin_%"].mean()
print(avg_margin_fixed)
```
