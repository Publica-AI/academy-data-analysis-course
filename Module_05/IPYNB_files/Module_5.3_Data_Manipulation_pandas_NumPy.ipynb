## Part 1 — Filtering: one condition

**Excel task:** Filter, keep only rows where Sales is greater than 100,000.

```python
df[df["Sales"] > 100000]
```

`df["Sales"] > 100000` checks every row and produces `True` or `False`.
Wrapping it in `df[...]` keeps only the rows marked `True`. Same filter-arrow
logic, the condition just moves from a dropdown menu into one line of code.

## Part 2 — Filtering: multiple conditions

`&` means both conditions must be true. `|` means either can be true. Every
condition needs its own parentheses, this is not optional in pandas.

```python
df[(df["Sales"] > 100000) & (df["Region"] == "South West")]
```

```python
df[(df["Region"] == "South West") | (df["Region"] == "North West")]
```

**Excel parallel:** the same result as Excel's Filter with two criteria
applied at once, or Custom Filter with AND/OR, except both filters are visible
in the code itself, not hidden in a dialog box.

## Part 3 — Sorting

**Excel task:** Sort, Largest to Smallest on the Sales column.

```python
df.sort_values("Sales", ascending=False)
```

- `ascending=False` sorts largest to smallest.
- `ascending=True`, or leaving it out, sorts smallest to largest.
- `sort_values()` never changes `df` itself, it returns a new sorted view.

Misreading `ascending` is the single most common sorting mistake, always check
the first two rows after sorting.

## Part 4 — Grouping

**Group → Select → Aggregate.** This is exactly a PivotTable: drag Region into
Rows, drag Sales into Values, Excel groups and sums for you the same way
`groupby` does.

```python
df.groupby("Region")["Sales"].sum()
```

- `groupby("Region")` groups every row by its Region value.
- `["Sales"]` selects the column to work on inside each group.
- `.sum()` aggregates each group down to one number.
- The group labels become the new index, sorted alphabetically by default.

South West alone outsells the other two regions combined, from ten of the
twenty transactions.

Run each of these on the Sales column and see what business question it

## Part 6 — Multiple aggregations at once: `.agg()`

Instead of three separate `groupby` lines, `.agg([...])` runs all three at
once, in one table.

```python
df.groupby("Region")["Sales"].agg(["sum", "mean", "count"])
```

South West has the most transactions and the highest total, but the lowest
average transaction size. Three numbers tell a fuller story than one.

## Part 8 — NumPy essentials for analysts

NumPy is the numerical engine pandas is built on. Every pandas column is a
NumPy array underneath, which is why dtypes are named `int64` and `float64`,
NumPy's names, not pandas' own.

**Conditional calculation:**

```python
df["Sales_Tier"] = np.where(df["Sales"] > 100000, "High", "Low")
df["Sales_Tier"].value_counts()
```

**Rounded math:**

```python
df["Estimated_Units"] = np.round(df["Sales"] / df["Price"], 1)
df[["Branch", "Product", "Sales", "Price", "Estimated_Units"]].head(8)
```

**Notice:** a missing Price produces a missing `Estimated_Units`. `NaN`
spreads through arithmetic, exactly why Module 5.2 checks `isna()` first.

This is analyst-relevant NumPy only: conditional flags and clean rounding. No
advanced NumPy in this programme.

## Part 9 — Combining operations: one workflow

**Business question:** among high-value transactions (Sales above ₦100,000),
which region generated the most profit, and how many estimated units did it
move?

**1. Filter → 2. Calculate → 3. Sort → 4. Group + aggregate**, all in one
short pipeline.

```python
workflow = df[df["Sales"] > 100000].copy()
workflow["Estimated_Units"] = np.round(workflow["Sales"] / workflow["Price"], 1)
workflow = workflow.sort_values("Sales", ascending=False)

result = workflow.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Total_Estimated_Units=("Estimated_Units", "sum"),
    Transactions=("Sales", "count"),
)
result
```

South West leads on every metric among high-value transactions: highest
total sales, highest total profit, and the most estimated units moved.

## Part 10 — Excel → pandas translation

| Analysis task | Excel | pandas |
|---|---|---|
| Filter | Filter | Boolean condition, e.g. `df[df["Sales"] > 100000]` |
| Sort | Sort | `.sort_values()` |
| Pivot | PivotTable | `.groupby()` |
| Sum | `SUM()` | `.sum()` |
| Average | `AVERAGE()` | `.mean()` |
| Calculated field | Formula filled down | New column, e.g. `df["Profit"] = ...` |

Six familiar Excel tasks, six pandas equivalents. The thinking behind each one
hasn't changed.

## Part 11 — Reproducing an Excel analysis

**In Excel, you would:**

1. Build a PivotTable: Region in Rows, Sum of Sales in Values.
2. Add a second value field: Average of Profit Margin %.
3. Sort the PivotTable by Sales, largest to smallest.

**In pandas, you write:**

```python
df["Margin_%"] = (df["Profit"] / df["Sales"] * 100).round(1)
df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Avg_Margin=("Margin_%", "mean"),
).sort_values("Total_Sales", ascending=False)
```

> "Python is not replacing analytical thinking. It is changing how we
> execute the analysis."

Every region's margin comes out at exactly 15.0%, matching the PivotTable
exactly, and confirming the company's pricing policy is applied consistently.

## Common beginner mistakes

| Mistake | What goes wrong |
|---|---|
| Wrong column name | `df["Slaes"]` → `KeyError`. Check spelling and capitalisation against `df.columns`. |
| Missing quotation marks | `df[Sales]` → `NameError`. Column names are text, they need quotes. |
| `=` instead of `==` | `=` assigns a value, `==` compares one. Filtering always needs `==`. |
| Forgetting parentheses | `df.describe` (no parentheses) shows a method, not a result. Always call it: `df.describe()`. |
| Comparing the wrong thing | `"Sales" > 100000` compares text to a number and fails. Compare the column's values: `df["Sales"] > 100000`. |
| Sorting the wrong direction | `ascending=True` when you meant `False`. Always check the first two rows after sorting. |
| Grouping by the wrong column | `groupby("Branch")` when the question asks about Region. Reread the business question before writing the code. |
